```python
import pandas as pd
import os

def convert_csv_to_parquet(csv_file, parquet_file=None):
    """
    Converts a CSV file to a Parquet file.

    Args:
        csv_file (str): Path to the input CSV file.
        parquet_file (str, optional): Path to the output Parquet file. 
                                      If None, it will use the same name as the CSV file with .parquet extension.
    """
    try:
        # Read the CSV file into a pandas DataFrame
        df = pd.read_csv(csv_file)

        # Determine the output Parquet file name
        if parquet_file is None:
            base_name, _ = os.path.splitext(csv_file)
            parquet_file = base_name + ".parquet"

        # Write the DataFrame to a Parquet file
        df.to_parquet(parquet_file, index=False)  # index=False to avoid writing the index

        print(f"Successfully converted '{csv_file}' to '{parquet_file}'")

    except FileNotFoundError:
        print(f"Error: CSV file '{csv_file}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

csv_file_path = "raw.csv"  # Replace with your CSV file path
parquet_file_path = "raw.parquet" #replace with your desired parquet file path, or leave it none.

convert_csv_to_parquet(csv_file_path, parquet_file_path) #Or just convert_csv_to_parquet(csv_file_path)
```
